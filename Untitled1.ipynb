```python
%%writefile solution.py
from datetime import datetime
import multiprocessing
import random
import time

def print_time():
    time.sleep(random.random())
    print(datetime.now())

if __name__ == "__main__":
    processes = []
    for _ in range(3):
        p = multiprocessing.Process(target=print_time)
        processes.append(p)
        p.start()

        for p in processes:
            p.join()
```
